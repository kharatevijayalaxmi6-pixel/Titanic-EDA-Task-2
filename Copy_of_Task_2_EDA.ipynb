```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/Titanic_Cleaned (1).csv')
df.head()
```

```python
sns.boxplot(data=df, x='Survived', y='Age')

plt.title('Age Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Standardized Age')

plt.show()
```

### Age Distribution by Survival

1. The age distributions of survivors and non-survivors are fairly similar.
2. The median age is close for both groups.
3. Both groups contain some unusually high and low age values.
4. The considerable overlap between the two boxplots suggests that age alone does not show a very strong difference in survival in this dataset.

```python
sns.boxplot(data=df, x='Survived', y='Fare')

plt.title('Fare Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Standardized Fare')

plt.show()
```

### Fare Distribution by Survival

1. The median fare for survivors is higher than the median fare for non-survivors.
2. Survivors show a wider distribution of fare values.
3. Both groups contain some high-fare outliers.
4. The graph suggests that passengers who paid higher fares generally had better survival outcomes.
5. Fare appears to have a noticeable relationship with survival in the dataset.

```python
correlation = df.corr()

correlation['Survived'].sort_values(ascending=False)
```

### Correlation Analysis

1. Sex has the strongest correlation with Survival at 0.510.
2. Fare has a moderate positive correlation with Survival at 0.261.
3. Pclass has a moderate negative correlation with Survival at -0.273.
4. Age has a very weak negative correlation with Survival at -0.061.
5. SibSp has almost no linear correlation with Survival at 0.004.
6. Overall, Sex, Fare, and Pclass show the most noticeable relationships with Survival.

```python
plt.figure(figsize=(10, 8))

sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```

### Observations from Correlation Heatmap

1. Sex has the strongest correlation with Survival at approximately 0.51.
2. Fare has a positive correlation with Survival of approximately 0.26.
3. Pclass has a negative correlation with Survival of approximately -0.27.
4. Age has a very weak correlation with Survival of approximately -0.06.
5. SibSp has almost no correlation with Survival.
6. The heatmap provides a visual overview of relationships between all numerical variables.

```python
survival_by_sex = df.groupby('Sex')['Survived'].mean()

survival_by_sex
```

### Survival Rate by Sex

The survival rate for male passengers (Sex = 0) was approximately **17.14%**, while the survival rate for female passengers (Sex = 1) was approximately **68.83%**.

This shows a substantial difference in survival rates between males and females. Female passengers had a much higher survival rate than male passengers in this dataset.

```python
survival_by_class = df.groupby('Pclass')['Survived'].mean()

survival_by_class
```

### Survival Rate by Passenger Class

The survival rate for 1st-class passengers was approximately **56.84%**, for 2nd-class passengers approximately **47.56%**, and for 3rd-class passengers approximately **24.03%**.

The results show that 1st-class passengers had the highest survival rate, while 3rd-class passengers had the lowest survival rate. Overall, survival decreased as passenger class increased from 1st to 3rd class.

```python
overall_survival_rate = df['Survived'].mean()

overall_survival_rate
```

### Overall Survival Rate

The overall survival rate in the dataset is approximately **33.70%**.

This means that about one-third of the passengers in the cleaned dataset survived, while approximately two-thirds did not survive.

## Final Conclusion

The exploratory data analysis of the Titanic dataset shows that survival was strongly associated with **Sex** and **Passenger Class**. Female passengers had a much higher survival rate (**68.83%**) compared with male passengers (**17.14%**). Passengers in 1st class had the highest survival rate (**56.84%**), while 3rd-class passengers had the lowest (**24.03%**).

Fare also showed a positive relationship with survival, while Age showed only a very weak relationship. The overall survival rate in the cleaned dataset was approximately **33.70%**.

Overall, the analysis indicates that **Sex, Passenger Class, and Fare were important factors associated with survival in this dataset**.
